Q1. Analyze the distribution of earthquake magnitudes (mag) and explore how they vary across different earthquake types (type) using visualizations such as boxplots or violin plots.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load Dataset
df = pd.read_csv("preprocessed_earthquake_data.csv")

# Plot Boxplot of Magnitude for each Earthquake Type
plt.figure(figsize=(10,5))
df.boxplot(column='mag', by='type')
plt.xlabel("Earthquake Type")
plt.ylabel("Magnitude")
plt.title("Distribution of Earthquake Magnitudes Across Different Types")
plt.suptitle("")  # Removes extra title from pandas
plt.show()
```

The boxplot shows how earthquake magnitudes vary for different earthquake types.
From the plot, we can observe the spread of magnitudes and the median value for each type.
This helps us understand which earthquake types generally have stronger or weaker magnitudes.

Q2. Explore the relationship between earthquake depth (depth) and magnitude (mag). Is there a trend or correlation between these two features in the dataset?

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load Dataset
df = pd.read_csv("preprocessed_earthquake_data.csv")

# Scatter Plot: Depth vs Magnitude
plt.figure(figsize=(10,5))
plt.scatter(df['depth'], df['mag'])
plt.xlabel("Earthquake Depth")
plt.ylabel("Magnitude")
plt.title("Relationship Between Earthquake Depth and Magnitude")
plt.show()

# Calculate Correlation
correlation = df['depth'].corr(df['mag'])
print("Correlation between Depth and Magnitude:", correlation)
```

The scatter plot shows the relationship between earthquake depth and magnitude.
From the plot, we observe that the points are spread out without forming a strong
upward or downward trend. This indicates that there is no strong relationship
between depth and magnitude.

The correlation value is also close to 0, which confirms that earthquake depth
does not significantly influence the magnitude of the earthquake in this dataset.

Q3. Using the preprocessed temporal features (Year, Month, Day, Hour), investigate if there is any difference in earthquake frequency or magnitude over months or years. Visualize temporal trends clearly.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("preprocessed_earthquake_data.csv")

# Group by Year to count earthquakes per year
earthquakes_per_year = df.groupby('Year').size()

# Plot
plt.figure(figsize=(10,5))
plt.plot(earthquakes_per_year.index, earthquakes_per_year.values)
plt.xlabel("Year")
plt.ylabel("Number of Earthquakes")
plt.title("Earthquake Frequency Over Years")
plt.show()
```

The line graph shows how the number of earthquakes varies over different years.
We can observe that the frequency of earthquakes changes from year to year.
Some years have more earthquakes, while others have fewer.
This indicates that earthquake activity is not constant over time, and may be
influenced by natural geological factors.

Q4. Identify the most active geographical earthquake regions by plotting earthquake locations (latitude and longitude) on a map with color coding or size scaled by earthquake magnitude.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("preprocessed_earthquake_data.csv")

plt.figure(figsize=(8,6))
plt.scatter(df['longitude'], df['latitude'], s=df['mag']*5)  # size by magnitude
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Earthquake Locations (Size Represents Magnitude)")
plt.show()
```

The scatter plot shows earthquake locations based on latitude and longitude.
The size of each point represents the magnitude of the earthquake.
Regions where many points cluster closely are the most active geographical zones.
Larger circles in those clusters indicate stronger earthquakes in those areas.
