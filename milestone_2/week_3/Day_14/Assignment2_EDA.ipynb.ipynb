####Earthquake Frequency by Year

```python
import pandas as pd
import matplotlib.pyplot as plt

# Clean column names
df.columns = df.columns.str.strip().str.lower()

# Count earthquakes per year
year_counts = df['year'].value_counts().sort_index()

# Plot
plt.figure(figsize=(12,6))
plt.bar(year_counts.index, year_counts.values, color='skyblue')

plt.title('Earthquake Frequency by Year', fontsize=14)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Count', fontsize=12)

# Show every 5th or 10th year label only
plt.xticks(ticks=year_counts.index[::10], labels=year_counts.index[::10], rotation=45)

plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

3. A sharp increase is seen after the 1970s — possibly due to better global detection systems.

3. Recent years (2000–2024) show more consistent and higher frequency of recorded earthquakes.

##Task 4: Depth Analysis and Its Impact

#####Objective: Investigate how depth inuences seismic characteristics and damage.

###Tasks:

Create box plots: Depth vs. Magnitude (stratied by depth ranges: shallow, intermediate, deep).

Analyze depth distribution with histogram and statistics (mean, median, std).

Plot depth vs. damage_potential using scatter and 2D density plots.

Calculate mean damage_potential for different depth ranges.

Identify whether shallow or deep earthquakes cause more damage


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import Markdown, display

#  Load Dataset
df = pd.read_csv("/content/preprocessed_earthquake_data.csv")

# Ensure depth is numeric
df['depth'] = pd.to_numeric(df['depth'], errors='coerce')

# -------------------------------------------------------------
# Categorize Depth into Ranges

def depth_category(depth):
    if pd.isna(depth):
        return 'Unknown'
    elif depth < 70:
        return 'Shallow'
    elif 70 <= depth <= 300:
        return 'Intermediate'
    else:
        return 'Deep'

df['Depth_Category'] = df['depth'].apply(depth_category)
print("Depth Category Counts:\n", df['Depth_Category'].value_counts())

# -------------------------------------------------------------
# Analyze Depth Distribution (Histogram)

plt.figure(figsize=(8,5))
sns.histplot(df['depth'], bins=50, kde=True, color='orchid')
plt.title("Depth Distribution of Earthquakes", fontsize=14)
plt.xlabel("Depth (km)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

# -------------------------------------------------------------
# Box Plot — Depth Category vs Magnitude

plt.figure(figsize=(8,5))
sns.boxplot(
    x='Depth_Category',
    y='mag',
    data=df,
    order=['Shallow', 'Intermediate', 'Deep'],
    palette='Purples'
)
plt.title("Magnitude Distribution Across Depth Categories", fontsize=14)
plt.xlabel("Depth Category")
plt.ylabel("Magnitude")
plt.tight_layout()
plt.show()

# -------------------------------------------------------------
#  Scatter Plot — Depth vs Damage Potential

plt.figure(figsize=(8,6))
sns.scatterplot(
    x='depth',
    y='damage_potential',
    data=df.sample(2000, random_state=42),
    hue='mag',
    palette='viridis',
    alpha=0.7
)
plt.title("Depth vs Damage Potential (Colored by Magnitude)", fontsize=14)
plt.xlabel("Depth (km)")
plt.ylabel("Damage Potential")
plt.tight_layout()
plt.show()

# -------------------------------------------------------------
#  Depth-Based Summary Statistics

summary = df.groupby('Depth_Category')['damage_potential'].agg(['mean', 'median', 'std', 'count']).reset_index()
print("Depth-Based Damage Summary Table:")
print(summary)

# -------------------------------------------------------------
#  Step 7: Insights

display(Markdown("""
## Insights: Depth Analysis and Impact

1 **Box Plot:**
   - Shows that **shallow earthquakes** occur most frequently and can have a **wide range of magnitudes**.
   - **Intermediate and deep earthquakes** are less frequent but sometimes reach high magnitudes.

2 **Histogram:**
   - Most earthquakes are concentrated **below 100 km**, confirming that shallow crustal quakes dominate globally.

3 **Scatter Plot:**
   - Clear trend: **deeper earthquakes have lower damage potential**, while **shallow quakes** show higher impact at similar magnitudes.

4 **Summary Table:**
   - The **mean and median damage potential** is **highest for shallow** quakes.
   - **Standard deviation (std)** indicates greater variation in shallow earthquakes damage range.

5 **Conclusion:**
   - **Shallow earthquakes (<70 km)** cause the most damage.
   - **Depth and damage potential are inversely related** — the deeper the earthquake, the less destructive it tends to be.

 **Deliverables Covered:**
✔ Box plots
✔ Histograms
✔ Scatter plots
✔ Depth-based summary tables
"""))
```

##Task 5: Categorical Feature Analysis

####Objective: Explore relationships between earthquake types, magnitude types, and status.


##Tasks:

Analyze value counts for magType, type, and status columns

Create bar charts showing frequency distribution of each categorical variable

Generate grouped bar charts: magType vs. mean magnitude and damage_potential

Create cross-tabulation (contingency tables) between categorical variables

Visualize earthquake type distribution and its association with damage_potential

Calculate Cramér's V statistics for categorical associations

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from scipy.stats import chi2_contingency
from IPython.display import Markdown, display

#  Load Dataset
df = pd.read_csv("/content/preprocessed_earthquake_data.csv")

# Clean and ensure categorical columns exist
cat_cols = ['type', 'magType', 'status']
for col in cat_cols:
    df[col] = df[col].astype(str).str.strip()

# -------------------------------------------------------------
# Analyze Value Counts (Frequency Distribution)

for col in cat_cols:
    print(f"\n🔹 Value Counts for {col}:")
    print(df[col].value_counts())

# -------------------------------------------------------------
#  Create Bar Charts for Each Categorical Feature

plt.figure(figsize=(16,4))
for i, col in enumerate(cat_cols):
    plt.subplot(1,3,i+1)
    sns.countplot(data=df, x=col, palette='mako')
    plt.title(f"Frequency of {col.capitalize()}")
    plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# -------------------------------------------------------------
# Grouped Bar Charts — Category vs Mean Magnitude & Damage Potential

group_stats = df.groupby(['type'])[['mag', 'damage_potential']].mean().reset_index()

plt.figure(figsize=(8,5))
sns.barplot(data=group_stats, x='type', y='mag', color='violet', label='Mean Magnitude')
sns.barplot(data=group_stats, x='type', y='damage_potential', color='purple', alpha=0.5, label='Damage Potential')
plt.title("Mean Magnitude and Damage Potential by Earthquake Type", fontsize=13)
plt.ylabel("Value")
plt.xlabel("Earthquake Type")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# -------------------------------------------------------------
#   Contingency Tables (Cross-tabulations)

type_mag_ct = pd.crosstab(df['type'], df['magType'])
status_type_ct = pd.crosstab(df['status'], df['type'])

print("\n Contingency Table: Type vs Magnitude Type")
print(type_mag_ct)
print("\n Contingency Table: Status vs Type")
print(status_type_ct)

# -------------------------------------------------------------
#  Step 6: Calculate Cramer's V for Associations

def cramers_v(confusion_matrix):
    chi2 = chi2_contingency(confusion_matrix)[0]
    n = confusion_matrix.sum().sum()
    phi2 = chi2 / n
    r, k = confusion_matrix.shape
    phi2corr = max(0, phi2 - ((k-1)*(r-1))/(n-1))
    rcorr = r - ((r-1)**2)/(n-1)
    kcorr = k - ((k-1)**2)/(n-1)
    return np.sqrt(phi2corr / min((kcorr-1), (rcorr-1)))

cramer_results = {
    'Type vs MagType': cramers_v(type_mag_ct),
    'Status vs Type': cramers_v(status_type_ct)
}

print("\n Cramer’s V Association Strengths:")
for k, v in cramer_results.items():
    print(f"{k}: {v:.3f}")

# -------------------------------------------------------------
#  Step 7: Visualize Cramer's V Matrix (Heatmap)

cramer_matrix = pd.DataFrame.from_dict(cramer_results, orient='index', columns=['Cramer_V'])
plt.figure(figsize=(5,2.5))
sns.heatmap(cramer_matrix, annot=True, cmap='Purples', fmt=".3f")
plt.title("Cramer's V Association Strengths")
plt.show()

# -------------------------------------------------------------
#  Step 8: Insights & Summary

display(Markdown("""
##  Insights: Categorical Feature Analysis

 **Frequency Distribution:**
   - Majority of earthquakes are of type *earthquake* or *quarry blast*.
   - Most magnitudes are recorded under *mb* or *ml* types.

 **Grouped Bar Chart:**
   - Some types (e.g., *explosion*, *nuclear test*) show lower average magnitudes but high damage potential.
   - Natural earthquakes have consistent magnitude and damage profiles.

 **Contingency Tables:**
   - Reveal frequency relationships between *status*, *type*, and *magType* combinations.

 **Cramer’s V Matrix:**
   - Indicates **moderate correlation** between *earthquake type* and *magnitude type*.
   - *Status* and *type* have **weak association**, meaning reporting status doesn’t affect type much.

**Deliverables Covered:**
✔ Bar charts
✔ Contingency tables
✔ Cramer’s V matrices
✔ Frequency distributions
"""))
```
